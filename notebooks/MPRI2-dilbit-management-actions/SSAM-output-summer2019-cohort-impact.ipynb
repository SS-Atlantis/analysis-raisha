50 year Atlantis results for the 2019-07-03:summer_Fraser_strong_winds scenario

```python
import os
import re
import xarray as xr
import numpy as np
import itertools
import pandas as pd
import geopandas as gpd
import matplotlib.cm as cm
from pathlib import Path
from datetime import timedelta
import matplotlib.pyplot as plt
import ssam_groups as groups
```

```python
# Read in salish sea atlantis output files.
control_file = "/ocean/rlovindeer/MOAD/analysis-raisha/SSmodel_outputs/Spills/MPRI2/control-AtlantisTrunk/outputSalishSea.nc"
control = xr.open_dataset(str(control_file), decode_cf=True)
time = np.ma.filled(control.variables['t'])
```

```python
scenario_root = Path('/ocean/rlovindeer/MOAD/analysis-raisha/SSmodel_outputs/Spills/MPRI2/')
scenario_paths = sorted([p for p in scenario_root.glob('fiftyyear_5b*/outputSalishSea.nc')])
for path in scenario_paths:
    print(path.parent.stem)
```

```python
scenario_datasets = [xr.open_dataset(scen,decode_cf=True) for scen in scenario_paths]
```

```python
# time after burn-in
start = 0
end = time.size-1
```

```python
line_styles = itertools.cycle(('-', '--', '-.', ':'))#,(5, (10, 3)),
#    (0, (3, 1, 1, 1)),(0, (1, 1)),(0, (3, 10, 1, 10, 1, 10))))
```

```python
species = 'Small_pelagic_fish'
numCohorts = groups.cohorts[species]
boxes = groups.salish_sea
start = 312
end = 432

for scenario in scenario_datasets:
    fig, ax = plt.subplots(figsize = (14,5))
    for cohort in range (1, numCohorts+1):
        new_species = species + str(cohort)
        pelagic_control_tbl = np.ma.filled(control.variables[new_species+'_Nums'][start:end,boxes,0:6], np.nan) 
        pelagic_control = pelagic_control_tbl.sum(axis=(1,2))
        #ax.plot(control.t[start:end], pelagic_control, 'k', linewidth = 2, label='control')

        pelagic_oiled_tbl = np.ma.filled(scenario.variables[new_species+'_Nums'][start:end,boxes,0:6], np.nan)
        pelagic_oiled = pelagic_oiled_tbl.sum(axis=(1,2))
        
        ratio = (pelagic_oiled / pelagic_control - 1) * 100
        ax.plot(scenario.t[start:end], ratio, linewidth = 4, label='cohort'+str(cohort), linestyle=':') #next(line_styles))
            
    plt.ylabel('percent change in numbers', fontsize=12)
    plt.minorticks_on()
    plt.grid()
    plt.legend()
    ax.set_title(species)
fig.savefig('50y-cohort'+species+str(end)+'.png',bbox_inches='tight',dpi=300)
```
