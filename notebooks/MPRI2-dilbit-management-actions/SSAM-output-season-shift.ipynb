Disentangling the influence of seasonality within Atlantis on biological impact by forcing the Parcels run from 2019-07-03 (summer, free_PAH simulation) six months earlier, during winter in Atlantis.

```python
import os
import xarray as xr
import numpy as np
import itertools
import warnings
import pandas as pd
import seaborn as sns
from pathlib import Path
import geopandas as gpd
import matplotlib.cm as cm
import matplotlib.colors as colors
import matplotlib.pyplot as plt
import ssam_groups as groups
from mpl_toolkits.axes_grid1 import make_axes_locatable

warnings.simplefilter('ignore')
```

```python
forcing_palette = sns.color_palette([
'#6699CC', #WINTER
'#EE7733',# SUMMER
])
simulation_palette = sns.color_palette([
'#EE7733',# SUMMER
'#6699CC', #WINTER
])
```

### Scenario comparison across fished groups

```python
df = pd.read_csv("/ocean/rlovindeer/MOAD/analysis-raisha/SSmodel_outputs/Spills/MPRI2/data_season_shift.csv")
```

```python
plt.figure(figsize=(10, 8))
plt.grid()
ax = sns.violinplot(df, x="description", y="percent_change",palette=forcing_palette) 
plt.ylabel('Percent (%) Change', fontsize = 14)
#plt.xticks(rotation=90)
plt.title('Impact of seasonal shift in dilbit-PAH forcing', fontsize = 14)
plt.axhline(0, color='k'); 
```

```python
plt.figure(figsize=(10, 8)) 
plt.grid()
ax=sns.violinplot(df, x="description", y="percent_change",
    palette=forcing_palette, hue='description', fill=False, zorder=10, legend=False) 
sns.stripplot(data=df, x="description", palette=simulation_palette,
    y="percent_change", hue='sim_description', alpha=0.5, ax=ax)
plt.ylabel('Percent (%) Change', fontsize = 14)
plt.title('Impact of seasonal shift in dilbit-PAH forcing', fontsize = 14)
plt.axhline(0, color='k'); #linestyle='--'
plt.xlabel(None)
plt.xticks(fontsize=14)
plt.legend(fontsize=12);
#plt.savefig('Fig-results-scenarios.png',bbox_inches='tight',dpi=300);
```

The season of the PAH release in Parcels (dot colours) is the stronger determinant of the impact across the majority of the biological functinal groups, with summer releases (orange) having a stronger impact than winter releases (blue) regardless of the base season in Atlantis (colour of the ouline).

The only guild affected by the season of the Atlantis base model is the planktonic guild. Picoplankton are more affected in the Atlantis winter regardless of the Parcels season. Pelagic bacteria and microzoo are more variable across all the combinations tested.
